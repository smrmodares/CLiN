# Ngram

```python
# ======== hazm setup ========
normalizer = hazm.Normalizer()
lemmatizer = hazm.Lemmatizer()
tokenizer = hazm.WordTokenizer()

# ======== stopwords hazm ========
persian_stopwords = set(hazm.stopwords_list())

# ======== baseline: unigram diversity (بدون stopwords) ========
def elaboration_baseline(text):
    if not isinstance(text, str) or text.strip() == "":
        return 0
    
    text = normalizer.normalize(text)
    tokens = tokenizer.tokenize(text)
    
    # حذف stopwords
    content_tokens = [t for t in tokens if t not in persian_stopwords]
    
    # unigram diversity
    return len(set(content_tokens))


# ======== query-wise aggregation ========
def query_wise_elaboration(df):
    results = []
    for query, group in df.groupby("Query"):
        baseline_scores = []
        
        for s in group["Sentence"]:
            baseline_scores.append(elaboration_baseline(s))
        
        results.append({
            "Topic": query,
            "Elaboration (unigram diversity, no stopwords)": sum(baseline_scores) / len(baseline_scores),
            "count_sentences": len(group)
        })
    
    return pd.DataFrame(results)
```

```python
def sentence_wise_elaboration(df):
    results = []
    for idx, row in df.iterrows():
        text = row["Sentence"]
        
        # baseline n-gram diversity
        baseline_score = elaboration_baseline(text)
        
        results.append({
            "Sentence": text,
            "Elaboration_baseline": baseline_score,
        })
    return pd.DataFrame(results)
```
